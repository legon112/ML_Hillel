```python
import gymnasium as gym
import imageio
import ale_py
from huggingface_sb3 import load_from_hub
from stable_baselines3 import PPO
from stable_baselines3.common.env_util import make_atari_env
from stable_baselines3.common.vec_env import VecFrameStack
```

Хочу спробувати декілька моделей, адже дуже цікаво подивитись реалізації. 

Почну з Lunar Lander

Завантажую модель з репозиторію

```python
checkpoint_lunar = load_from_hub("araffin/ppo-LunarLander-v2", "ppo-LunarLander-v2.zip")
```

Завантажую налаштування моделі

```python
model_lunar = PPO.load(checkpoint_lunar)
```

Роблю середовище для моделі, буду зберігати кадри

```python
env_lunar = gym.make("LunarLander-v3", render_mode="rgb_array")
obs, info = env_lunar.reset()
```

```python
max_steps = 600
```

```python
frames_lunar = []
step = 0
end_flag = False

# Виконується поки не пройдуть усі кроки та гра після цього не закінчиться
while not end_flag:
    step += 1
    
    # Передаю моделі данні про поточний стан та чекаю її дію
    action, _states = model_lunar.predict(obs, deterministic=True)
    
    # Виконую дію в середовищі та отримую нові дані
    obs, reward, terminated, truncated, info = env_lunar.step(action)
    
    # Роблю кадр та дадою у список
    frame = env_lunar.render()
    frames_lunar.append(frame)
    
    # Якщо гра закінчилась або макс кроків, то перезапускаю середовище
    if terminated or truncated:
        # Якщо пройшло макс кроків, то ставлю прапорець закінчення гри
        if (step >= max_steps):
            end_flag = True
        obs, info = env_lunar.reset()

env_lunar.close()
```

Генерую з кадрів гифку

```python
output_file = "lunar_lander.gif"
imageio.mimsave(output_file, frames_lunar, fps=30)
```

**Тепер спробую CartPole**

```python
checkpoint_cart = load_from_hub("HumanCompatibleAI/ppo-seals-CartPole-v0", "ppo-seals-CartPole-v0.zip")
model_cart = PPO.load(checkpoint_cart)
```

```python
env_cart = gym.make("CartPole-v0", render_mode="rgb_array")
obs, info = env_cart.reset()

frames_cart = []

for step in range(max_steps):
    action, _states = model_cart.predict(obs, deterministic=True)
    
    obs, reward, terminated, truncated, info = env_cart.step(action)
        
    frame = env_cart.render()
    frames_cart.append(frame)
    
    if terminated or truncated:
        obs, info = env_cart.reset()

env_cart.close()
```

```python
output_file = "cartpole.gif"
imageio.mimsave(output_file, frames_cart, fps=30)
```

**Чудово! Останнє, що спробую, буде SpaceInvaders**

```python
checkpoint_space = load_from_hub(repo_id="ThomasSimonini/ppo-SpaceInvadersNoFrameskip-v4", filename="ppo-SpaceInvadersNoFrameskip-v4.zip")
model_space = PPO.load(checkpoint_space)
```

Для цієї моделі треба буде інша функція для емуліції атарі + треба слеювати кадри

```python
gym.register_envs(ale_py)

env_space = make_atari_env("ALE/SpaceInvaders-v5", n_envs=1, env_kwargs={"render_mode": "rgb_array"})

env_space = VecFrameStack(env_space, n_stack=4)

obs = env_space.reset()
```

```python
frames_space = []
max_steps_space = 2000
lifes = 3

for step in range(max_steps_space):
    action, _states = model_space.predict(obs, deterministic=True)
    
    obs, reward, done, info = env_space.step(action)
        
    frame = env_space.render()
    frames_space.append(frame)

    if done[0]:
        if info[0]["lives"] == 0 or (lifes == info[0]["lives"]):
            print(f"Кінець гри на кроці {step}! Набрано очок: {info[0]['episode']['r']}")
            lifes = 3
        else:
            lifes -= 1
        
env_space.close()
```

```python
output_file = "SpaceInvaders.gif"
imageio.mimsave(output_file, frames_space, fps=30)
```

Вийшло, максимальна кількість балів 600!
