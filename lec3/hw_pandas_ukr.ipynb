```python
import pandas as pd
```

```python
data = pd.read_csv('adult.data.csv')
data.head()
```

**1. Скільки чоловіків и жінок (ознака *sex*) представлено в цьому наборі даних?**

```python
sex_counts = data['sex'].value_counts()
print(f"Кількість чоловіків у наборі данних: {sex_counts['Male']} осіб")
print(f"Кількість жінок у наборі данних: {sex_counts['Female']} осіб")
```

**2. Який середній вік (ознака *age*) жінок?**

```python
female_mask = data['sex'] == 'Female'
female_mean_age = data[female_mask]['age'].mean()
print(f"Середній вік жінок: {female_mean_age:.1f} років")
```

**3. Яка доля громадян Німеччини (ознака *native-country*)?**

```python
statistic_by_nat_country = data['native-country'].value_counts(normalize=True)
german_proportion = statistic_by_nat_country['Germany']
print(f"Доля громадян німеччини: {german_proportion:.4f}")
```

**4-5. Які средні значення та средньоквадратичні відхилення віку тих, хто отримує більше 50K на рік (ознака *salary*) і тих, хто отримує менше 50K на рік?**

```python
statistic_age = data.groupby(['salary'])['age'].describe()

mean_age = statistic_age['mean']['<=50K']
std_age = statistic_age['std']['<=50K']
print(f"Статистика тих, хто отримує менше 50K на рік")
print(f"Cреднє значення: {mean_age:.1f}")
print(f"Средньоквадратичне відхилення: {std_age:.1f}")

mean_age = statistic_age['mean']['>50K']
std_age = statistic_age['std']['>50K']
print(f"\nСтатистика тих, хто отримує більше 50K на рік")
print(f"Cреднє значення: {mean_age:.1f}")
print(f"Средньоквадратичне відхилення: {std_age:.1f}")
```

**6. Чи правда, що люди, які отримують більше 50k, мають як мінімум вищу освіту? (ознака *education – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters* или *Doctorate*)**

Треба перевірити на наявність освіти потрібну вибірку, результат буде 1 або 0 (True або False), якщо середнє значення буде 1, то усі мають вищу освіту

```python
salary_mask = data['salary'] == '>50K'
education_list = ["Bachelors", "Prof-school", "Assoc-acdm", "Assoc-voc", "Masters", "Doctorate"]
result = data[salary_mask]['education'].isin(education_list).mean() == 1.0
print(f"Чи всі люди, які отримують більше 50K на рік мають вищу освіту?\nВідповідь: {result}")
```

**7. Порахуйте статистику віку для кожної раси (ознака *race*) і кожної статі. Використайте *groupby* и *describe*. Знайдіть таким сином максимальний вік чоловіків раси *Amer-Indian-Eskimo*.**

```python
age_statistic_by_race_sex = data.groupby([ 'race', 'sex'])['age'].describe()
print("Статистика віку для кожної та статі раси:")
print(age_statistic_by_race_sex)

max_age_of_race = int(age_statistic_by_race_sex['max']['Amer-Indian-Eskimo']['Male'])
print(f"\nМаксимальний вік серед чоловіків раси Amer-Indian-Eskimo: {max_age_of_race} років")
```

**8. Серед кого більша доля багатіших (>50K): серед одружених чи не одружених чоловіків (ознака *marital-status*)? За одружених вважаємо тих, у кого *marital-status* починається із *Married* (Married-civ-spouse, Married-spouse-absent или Married-AF-spouse), всі інші - не одружені.**

```python
sex_mask = data['sex'] == 'Male'
task_8 = data[sex_mask].copy()
task_8['marital-status']= task_8['marital-status'].str.startswith('Married')
```

Після вашого зауваження я передевився завдання та не зовсім розумію, яку саме ціль нам треба знайти, тому пропоную 2 варіанти інтерпретації завдання:

1. групування за сімейним станом та вирахування долі долі багатіших в кожній групі

```python
task_8_grouped = task_8.groupby(['marital-status'])['salary'].value_counts(normalize=True)
result = task_8_grouped[True]['>50K'] >= task_8_grouped[False]['>50K']
answer = "" if result else "не "

print(f"В середені груп одружених та неодружених чоловіків, група {answer}одружених має більшу частку тих, хто отримує більше 50 тисяч")
```

2. групування за зарплатньою та вирахування долі одружених та неодружених в кожній групі

```python
task_8_grouped = task_8.groupby(['salary'])['marital-status'].value_counts(normalize=True)
result = task_8_grouped['>50K'][True] >= task_8_grouped['>50K'][False]
answer = "" if result else "не "

print(f"Серед тих хто заробляє більше 50 тисяч, більша частка {answer}одружених чоловіків.")
```

**9. Яку максимальну кількість годин в тиждень працює людина (ознака *hours-per-week*)? Скільки людей працюють таку кількість годин і який відсоток серед них багатих?**

```python
max_hours = data['hours-per-week'].max()
max_hours_mask = (data['hours-per-week'] == max_hours)
max_hours_workers = data[max_hours_mask]

count_max_hours_workers = max_hours_workers.shape[0]

percent_of_rich = max_hours_workers['salary'].value_counts(normalize=True)['>50K'] * 100

print(f"Максимальна кількість годин роботи на тиждень: {max_hours} годин")
print(f"Кількість людей, які працюють таку кількість годин на тиждень: {count_max_hours_workers} осіб")
print(f"Відсоток людей, які працюють максимальну кількість годин на тиждень і отримують більше 50 тисяч: {percent_of_rich:.2f}%")
```

**10. Порахуйте середню тривалість роботи на тиждень (*hours-per-week*) серед багатих і бідних (*salary*) для кожної країни (*native-country*).**

```python

mean_hours_by_country_for_salary = data.groupby(['native-country', 'salary'])['hours-per-week'].describe()['mean']
print("Cередня тривалість роботи на тиждень серед багатих і бідних для кожної країни:")
print(mean_hours_by_country_for_salary)
```
