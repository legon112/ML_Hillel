```python
import keras
```

```python
(train_images, train_labels), (test_images, test_labels) = keras.datasets.mnist.load_data()
```

Нормалізація від 0 до 1

```python
train_images, test_images = train_images / 255.0, test_images / 255.0
```

Створю вхід

```python
inputs = keras.Input(shape=(28, 28))
```

Створю вхідний шар

```python
x = keras.layers.Flatten()(inputs)
```

Тепер створю перший скритий шар

```python
x = keras.layers.Dense(128, activation='relu')(x)
```

Вихідний шар, у такого шару активатор завжди softmax

```python
outputs = keras.layers.Dense(10, activation='softmax')(x)
```

Склам усе в модель

```python
model = keras.Model(inputs=inputs, outputs=outputs)
```

Тепер скомпілюю

```python
model.compile(optimizer='adam',
			loss='sparse_categorical_crossentropy',
			metrics=['accuracy'])
```

```python
model.summary()
```

Тепер тренування

```python
history = model.fit(
    train_images,      
    train_labels, 
    epochs=5,          
    batch_size=64,     
    validation_split=0.1
)
```

Перевірка результатів

```python
test_loss, test_acc = model.evaluate(test_images, test_labels)
print(f"Test accuracy: {test_acc}")
```

Дуже хороший результат, спробую, зменшити кількість епох до двох

```python
inputs = keras.Input(shape=(28, 28))
x = keras.layers.Flatten()(inputs)
x = keras.layers.Dense(128, activation='relu')(x)
outputs = keras.layers.Dense(10, activation='softmax')(x)
model = keras.Model(inputs=inputs, outputs=outputs)
model.compile(optimizer='adam',
			loss='sparse_categorical_crossentropy',
			metrics=['accuracy'])
history = model.fit(
    train_images,      
    train_labels, 
    epochs=2,          
    batch_size=64,     
    validation_split=0.1
)
```

```python
test_loss, test_acc = model.evaluate(test_images, test_labels)
print(f"Test accuracy: {test_acc}")
```

Точність знизилась лише на 1%, тепер спробую 10 епох

```python
inputs = keras.Input(shape=(28, 28))
x = keras.layers.Flatten()(inputs)
x = keras.layers.Dense(128, activation='relu')(x)
outputs = keras.layers.Dense(10, activation='softmax')(x)
model = keras.Model(inputs=inputs, outputs=outputs)
model.compile(optimizer='adam',
			loss='sparse_categorical_crossentropy',
			metrics=['accuracy'])
history = model.fit(
    train_images,      
    train_labels, 
    epochs=10,          
    batch_size=64,     
    validation_split=0.1
)
```

```python
test_loss, test_acc = model.evaluate(test_images, test_labels)
print(f"Test accuracy: {test_acc}")
```

Точність збільшилась меньше ніж на 0.5%, спробую знову 5 епох, але додам ще один шар

```python
inputs = keras.Input(shape=(28, 28))
x = keras.layers.Flatten()(inputs)
x = keras.layers.Dense(128, activation='relu')(x)
x = keras.layers.Dense(64, activation='relu')(x)
outputs = keras.layers.Dense(10, activation='softmax')(x)
model = keras.Model(inputs=inputs, outputs=outputs)
model.compile(optimizer='adam',
			loss='sparse_categorical_crossentropy',
			metrics=['accuracy'])
history = model.fit(
    train_images,      
    train_labels, 
    epochs=5,          
    batch_size=64,     
    validation_split=0.1
)
```

```python
test_loss, test_acc = model.evaluate(test_images, test_labels)
print(f"Test accuracy: {test_acc}")
```

Результат збільшився на 0.2%. Спробую ще зробити через Sequential

```python
model = keras.Sequential()
model.add(keras.Input(shape=(28, 28)))
model.add(keras.layers.Flatten())
model.add(keras.layers.Dense(128, activation='relu'))
model.add(keras.layers.Dense(10, activation='softmax'))
model.compile(optimizer='adam',
			loss='sparse_categorical_crossentropy',
			metrics=['accuracy'])
history = model.fit(
    train_images,      
    train_labels, 
    epochs=5,          
    batch_size=64,     
    validation_split=0.1
)
```

```python
test_loss, test_acc = model.evaluate(test_images, test_labels)
print(f"Test accuracy: {test_acc}")
```
