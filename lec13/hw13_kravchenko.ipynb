```python
import keras
import time
import pandas as pd
```

```python
(train_images, train_labels), (test_images, test_labels) = keras.datasets.fashion_mnist.load_data()
```

```python
train_images, test_images = train_images / 255.0, test_images / 255.0
```

Буду виконувати через згортку і dense, почну з dense для тесту оптималізаторів

```python
optimizers = {
    'SGD': 'sgd',
    'RMSprop': 'rmsprop',
    'Adam': 'adam'
}
results_data = []
```

```python
def get_dense():
    inputs = keras.Input(shape=(28, 28))
    x = keras.layers.Flatten()(inputs)
    x = keras.layers.Dense(128, activation='relu')(x)
    outputs = keras.layers.Dense(10, activation='softmax')(x)
    model = keras.Model(inputs=inputs, outputs=outputs)
    return model
```

```python
for name, optimizer in optimizers.items():
    keras.utils.set_random_seed(42)
    
    model = get_dense()
    model.compile(optimizer=optimizer,
			loss='sparse_categorical_crossentropy',
			metrics=['accuracy'])
    
    print (f'Training with {name} optimizer')
    
    start_time = time.time()
    
    history = model.fit(
		train_images,      
        train_labels, 
        epochs=5,          
        batch_size=64
    )
    
    end_time = time.time()
    
    train_time = end_time - start_time

    test_loss, test_acc = model.evaluate(test_images, test_labels)
    
    results_data.append({
            "Model Type": 'Dense',
            "Optimizer": name,
            "Train Accuracy": f"{history.history['accuracy'][-1]:.4f}",
            "Train Loss": f"{history.history['loss'][-1]:.4f}",
            "Test Accuracy": f"{test_acc:.4f}",
            "Test Loss": f"{test_loss:.4f}",
            "Time": f"{train_time:.2f}"
        })
```

Зробимо результати

```python
def get_conv():
    inputs = keras.Input(shape=(28, 28, 1))
    x = keras.layers.Conv2D(32, (3, 3), activation='relu')(inputs)
    x = keras.layers.MaxPooling2D((2, 2))(x)
    x = keras.layers.Flatten()(x)
    outputs = keras.layers.Dense(10, activation='softmax')(x)
    return keras.Model(inputs=inputs, outputs=outputs)
```

```python
for name, optimizer in optimizers.items():
    keras.utils.set_random_seed(42)
    
    model = get_conv()
    model.compile(optimizer=optimizer,
			loss='sparse_categorical_crossentropy',
			metrics=['accuracy'])
    
    print (f'Training with {name} optimizer')
    
    start_time = time.time()
    
    history = model.fit(
		train_images,      
        train_labels, 
        epochs=5,          
        batch_size=64
    )
    
    end_time = time.time()
    train_time = end_time - start_time

    test_loss, test_acc = model.evaluate(test_images, test_labels)
    
    results_data.append({
            "Model Type": 'Conv2D',
            "Optimizer": name,
            "Train Accuracy": f"{history.history['accuracy'][-1]:.4f}",
            "Train Loss": f"{history.history['loss'][-1]:.4f}",
            "Test Accuracy": f"{test_acc:.4f}",
            "Test Loss": f"{test_loss:.4f}",
            "Time": f"{train_time:.2f}"
        })
```

```python
df_results = pd.DataFrame(results_data)
df_results
```

Найточніша на цьому датасеті виявилась згорткова модель з оптімізатором RMSprop, так ще ї найшвидше серед згорткових моделей.

Трішки гірше впоралась згорткова з Адамом та вона виявилась трошки довша.

Найгірше та найдовше з усіх впоралась згорткова з SGD оптимізацією.

Серед Dense найкраще та з усіх найшвидше впоралась оптимізація Adam. Загалом Dense працювала швидке ніж згорткова

Тепер я хочу погратися з налаштуванням оптимізаторів, візьму Dense RMSprop та Conv2D Adam

```python
learning_rates = [0.3, 0.1, 0.03, 0.01, 0.003, 0.001]
results = []
```

```python
for i in learning_rates:
    
    model = get_dense()
    optimizer_dense = keras.optimizers.RMSprop(learning_rate=i)
    model.compile(optimizer=optimizer_dense,
			loss='sparse_categorical_crossentropy',
			metrics=['accuracy'])
    start_time = time.time()
    history_dense = model.fit(
		train_images,      
        train_labels, 
        epochs=5,          
        batch_size=64
    )
    end_time = time.time()
    train_time = end_time - start_time
    test_loss, test_acc = model.evaluate(test_images, test_labels)
    results.append({
        "Model Type": 'Dense',
        "Optimizer": f"RMSprop (lr={i})",
        "Train Accuracy": f"{history_dense.history['accuracy'][-1]:.4f}",
        "Train Loss": f"{history_dense.history['loss'][-1]:.4f}",
        "Test Accuracy": f"{test_acc:.4f}",
        "Test Loss": f"{test_loss:.4f}",
        "Time": f"{train_time:.2f}"
    })
    
    model = get_conv()
    optimizer_conv = keras.optimizers.Adam(learning_rate=i)
    model.compile(optimizer=optimizer_conv,
			loss='sparse_categorical_crossentropy',
			metrics=['accuracy'])
    start_time = time.time()
    history_conv = model.fit(
		train_images,      
        train_labels, 
        epochs=5,          
        batch_size=64
    )
    end_time = time.time()
    train_time = end_time - start_time
    test_loss, test_acc = model.evaluate(test_images, test_labels)
    
    results.append({
        "Model Type": 'Conv2D',
        "Optimizer": f"Adam (lr={i})",
        "Train Accuracy": f"{history_conv.history['accuracy'][-1]:.4f}",
        "Train Loss": f"{history_conv.history['loss'][-1]:.4f}",
        "Test Accuracy": f"{test_acc:.4f}",
        "Test Loss": f"{test_loss:.4f}",
        "Time": f"{train_time:.2f}"
    })
    
```

```python
df_results = pd.DataFrame(results)
df_results.sort_values(by=['Test Accuracy'], inplace=True)
df_results
```

Бачимо, що набагато краще вийшло у згортки при 0.003 ступені навчання, але вона ж і найдовше навчалась.

Видно що має сенс ставити lr < 0.1, та на цьому датасеті в тисячних lr веде себе найкраще
